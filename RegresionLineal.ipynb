# REGRESIÓN LINEAL

```python
# Importamos las librerías
import pandas as pd
from sklearn.linear_model import LinearRegression
```

```python
# Creamos el conjunto de datos
df = pd.DataFrame({
    "horas": [1, 2, 3, 4, 5],
    "nota": [2, 3, 5, 4, 6] 
})

X = df[["horas"]]
y = df[["nota"]]
print(X)
```

```python
# Creamos nuestro modelo
modelo = LinearRegression()
modelo.fit(X,y)
```

```python
# Imprimimos el interceptor y el coef
print("Interceptor: ", modelo.intercept_)
print("Pendiente: ", modelo.coef_[0])
```

```python
# Realizamos la nueva predicción
prediccion = modelo.predict([[6]])
print("Predicción: ", prediccion)
```

```python
# Graficar con matplotlib el modelo y verificar
# el nuevo valor corresponde a la predicción en la recta
import matplotlib.pyplot as plt

confgraf = df.plot(x='horas', y='nota', kind='scatter', color='purple', s=80, label='Datos iniciales')

horas_nueva = [[1], [2], [3], [4], [5], [6]]
prediccion_nueva = modelo.predict(horas_nueva)

plt.plot(horas_nueva, prediccion_nueva, color='orange', linewidth=2, label='Línea de regresión')

plt.scatter([6], [prediccion[0][0]], color='green', s=100, label='Predicción (6 hrs)')

plt.title('Gráfico de Regresión Lineal')
plt.xlabel('Horas')
plt.ylabel('Nota')
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()

plt.show()
```
